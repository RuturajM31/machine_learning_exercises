<h2 style="margin:0; font-size:24px; font-weight:800;">
5. What Are Hyperparameters?
</h2>

<p style="margin:8px 0 0 0; font-size:15px; color:#cbd5e1; line-height:1.6;">
Before we tune models, we must clearly understand what we are tuning.
</p>

</div>

<br>

<div style="border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px; font-family:Arial, sans-serif;">

<h3 style="margin:0 0 10px 0; color:#1e3a8a; font-size:20px;">
Simple Explanation
</h3>

<p style="margin:0; font-size:15px; line-height:1.7; color:#1f2937;">
A <b>hyperparameter</b> is a setting we choose <b>before</b> the model learns.
It controls how the model behaves during training.
</p>

<br>

<p style="margin:0; font-size:15px; line-height:1.7; color:#1f2937;">
Example: In a Random Forest, we can choose how many trees the model should build.
That setting is called <b>n_estimators</b>.
</p>

</div>

<br>

<div style="border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px; font-family:Arial, sans-serif;">

<h3 style="margin:0 0 10px 0; color:#166534; font-size:20px;">
Model Parameters vs Hyperparameters
</h3>

<ul style="margin:0; padding-left:20px; line-height:1.7; color:#1f2937; font-size:15px;">
<li><b>Model parameters</b> are learned by the model from the data.</li>
<li><b>Hyperparameters</b> are chosen by us before training starts.</li>
<li>The model learns parameters, but we tune hyperparameters.</li>
</ul>

</div>

<br>

<div style="border-left:6px solid #f59e0b; background:#fffbeb; padding:15px 18px; border-radius:8px; font-family:Arial, sans-serif;">

<h3 style="margin:0 0 10px 0; color:#92400e; font-size:20px;">
Business Meaning
</h3>

<p style="margin:0; font-size:15px; line-height:1.7; color:#1f2937;">
Hyperparameter tuning is important because the same model can perform very differently
with different settings. Good tuning can improve prediction quality, but too much tuning
can waste time and make the workflow harder to explain.
</p>

</div>

<h2 style="margin:0; font-size:23px; font-weight:800;">
9. Holdout Validation vs Cross-Validation
</h2>
<p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db; line-height:1.6;">
This section explains two common ways to check model performance before final testing.
</p>
</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#f9fafb; padding:15px 18px; border-radius:8px;">
<h3 style="margin:0 0 8px 0; color:#1e40af; font-size:19px;">
Holdout Validation
</h3>

<p style="margin:0; font-size:15px; line-height:1.7;">
In <b>holdout validation</b>, we split the data once.
One part is used for training, and one part is used for validation or testing.
</p>

<ul style="margin:10px 0 0 0; padding-left:20px; font-size:15px; line-height:1.7;">
<li><b>Good:</b> simple and fast.</li>
<li><b>Weakness:</b> the result depends strongly on one random split.</li>
</ul>
</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f8fff9; padding:15px 18px; border-radius:8px;">
<h3 style="margin:0 0 8px 0; color:#166534; font-size:19px;">
Cross-Validation
</h3>

<p style="margin:0; font-size:15px; line-height:1.7;">
In <b>cross-validation</b>, the data is split into several folds.
The model trains and validates multiple times, each time using a different validation fold.
</p>

<ul style="margin:10px 0 0 0; padding-left:20px; font-size:15px; line-height:1.7;">
<li><b>Good:</b> more stable than one split.</li>
<li><b>Weakness:</b> slower because the model trains many times.</li>
</ul>
</div>

<br>

<div style="border:1px solid #fde68a; border-left:6px solid #f59e0b; background:#fffbeb; padding:14px 18px; border-radius:8px;">
<p style="margin:0; font-size:15px; line-height:1.7; color:#78350f;">
<b>Key takeaway:</b> Holdout validation is faster, but cross-validation gives a more reliable score for tuning.
</p>
</div>

</div>

<h2 style="margin:0; font-size:23px; font-weight:800;">
10. Cross-Validation Flow
</h2>
<p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db; line-height:1.6;">
Cross-validation checks the model several times instead of trusting only one train-test split.
</p>
</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#f9fafb; padding:15px 18px; border-radius:8px;">

<h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">
Example: 5-Fold Cross-Validation
</h3>

<p style="margin:0 0 12px 0; font-size:15px; line-height:1.7;">
The dataset is split into <b>5 parts</b>.  
Each part gets one chance to become the validation set.
</p>

<div style="font-size:15px; line-height:1.8;">

<p style="margin:6px 0;"><b>Round 1:</b> Fold 1 = validation, Folds 2–5 = training</p>

```python
# ============================================================
# CELL 43 — GRID SEARCH WITH RANDOM FOREST
# ============================================================

# Start timer to measure Grid Search runtime.

grid_start_time = time.time()

# Build a Random Forest pipeline.

grid_rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            random_state=RANDOM_STATE,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)

# Assignment search space.
# "model__" means the parameter belongs to the model step inside the pipeline.

grid_param_grid = {
    "model__n_estimators": [50, 100, 200],
    "model__max_depth": [3, 5, None]
}

# GridSearchCV tests every parameter combination using 3-fold CV.

grid_search = GridSearchCV(
    estimator=grid_rf_pipeline,
    param_grid=grid_param_grid,
    scoring="f1",
    cv=3,
    n_jobs=-1
)


# Train Grid Search on the training data.
grid_search.fit(X_train, y_train)

# Calculate runtime.
grid_runtime = time.time() - grid_start_time

# Show main results.
print("Grid Search completed.")
print("Best parameters:", grid_search.best_params_)
print("Best CV F1 score:", round(grid_search.best_score_, 4))
print("Runtime seconds:", round(grid_runtime, 2))
```

```python
# ============================================================
# CELL 46 — RANDOM SEARCH WITH RANDOM FOREST
# ============================================================

# Start timer to measure Random Search runtime.

random_start_time = time.time()

# Build a Random Forest pipeline.
random_rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            random_state=RANDOM_STATE,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)

# Random Search space.
# It has more options than Grid Search, but only 10 random trials are tested.

random_param_grid = {
    "model__n_estimators": [50, 100, 150, 200, 250],
    "model__max_depth": [3, 5, 8, 10, None],
    "model__min_samples_leaf": [1, 2, 5]
}

# RandomizedSearchCV tests only 10 random combinations.
random_search = RandomizedSearchCV(
    estimator=random_rf_pipeline,
    param_distributions=random_param_grid,
    n_iter=10,
    scoring="f1",
    cv=3,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Train Random Search on the training data.
random_search.fit(X_train, y_train)

# Calculate runtime.
random_runtime = time.time() - random_start_time

# Show main results.
print("Random Search completed.")
print("Best parameters:", random_search.best_params_)
print("Best CV F1 score:", round(random_search.best_score_, 4))
print("Runtime seconds:", round(random_runtime, 2))
```

<h2 style="margin:0; font-size:23px; font-weight:800;">
17. Nested Cross-Validation
</h2>
<p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db; line-height:1.6;">
Nested Cross-Validation helps us avoid fooling ourselves after hyperparameter tuning.
</p>
</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#f9fafb; padding:15px 18px; border-radius:8px;">

<h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">
The problem
</h3>

<p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">
When we run <b>Grid Search</b> or <b>Random Search</b>, we test many hyperparameter combinations.

The search process keeps looking for the setting that performs best on validation folds.
</p>

<p style="margin:0; font-size:15px; line-height:1.7;">
This creates a risk: the selected setting may look strong because it fitted the validation process well,

not because it will truly perform well on new unseen data.
</p>

</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f8fff9; padding:15px 18px; border-radius:8px;">

<h3 style="margin16a34a; background:#f8fff9; padding:15px 18px; border-radius:8px;">

<h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">
Simple example
</h3>

<table style="width:100%; border-collapse:collapse; font-size:15px; line-height:1.6;">
<tr>
<th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Situation</th>
<th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">What happens</th>
<th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Risk</th>
</tr>

<tr>
<td style="padding:9px; border-bottom:1px solid #e5e7eb;"><b>Normal CV tuning</b></td>
<td style="padding:9px; border-bottom:1px solid #e5e7eb;">The same CV process is used to choose the best settings.</td>
<td style="padding:9px; border-bottom:1px solid #e5e7eb;">The score can become too optimistic.</td>
</tr>

<tr>
<td style="padding:9px;"><b>Nested CV</b></td>
<td style="padding:9px;">One CV loop tunes; another separate CV loop evaluates.</td>
<td style="padding:9px;">The final score is more honest.</td>
</tr>
</table>

</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #7c3aed; background:#f5f3ff; padding:15px 18px; border-radius:8px;">

<h3 style="margin:0 0 10px 0; color:#5b21b6; font-size:19px;">
How Nested CV works
</h3>

<table style="width:100%; border-collapse:collapse; font-size:15px; line-height:1.6;">
<tr>
<th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Loop</th>
<th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Job</th>
<th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Question it answers</th>
</tr>

<tr>
<td style="padding:9px; border-bottom:1px solid #e5e7eb;"><b>Inner loop</b></td>
<td style="padding:9px; border-bottom:1px solid #e5e7eb;">Runs Grid Search or Random Search.</td>
<td style="padding:9px; border-bottom:1px solid #e5e7eb;">Which hyperparameters are best?</td>
</tr>

<tr>
<td style="padding:9px;"><b>Outer loop</b></td>
<td style="padding:9px;">Tests the tuned model on data not used for tuning.</td>
<td style="padding:9px;">How good is the full tuning process?</td>
</tr>
</table>

</div>

<br>

<div style="border:1px solid #fde68a; border-left:6px solid #f59e0b; background:#fffbeb; padding:14px 18px; border-radius:8px;">
<p style="margin:0; font-size:15px; line-height:1.7; color:#78350f;">
<b>Key takeaway:</b> Nested CV does not only evaluate one model. 

It evaluates the complete tuning workflow, which makes the performance estimate more trustworthy.
</p>
</div>

</div>

```python
# ============================================================
# CELL 53 — NESTED CV WITH LARGER SEARCH SPACE
# ============================================================

# Start timer because Nested CV can be slow.
nested_start_time = time.time()

# Larger search space than the mini-demo.
# This gives the model more tuning options.
nested_param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [5, 8, None],
    "model__min_samples_leaf": [1, 2, 5],
    "model__max_features": ["sqrt", "log2"]
}

# Inner CV chooses the best hyperparameters.
inner_cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=RANDOM_STATE
)

# Outer CV checks the full tuning process fairly.
outer_cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=RANDOM_STATE
)

# Build the full pipeline.
nested_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            random_state=RANDOM_STATE,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)

# Inner loop: Grid Search.
nested_grid_search = GridSearchCV(
    estimator=nested_pipeline,
    param_grid=nested_param_grid,
    scoring="f1",
    cv=inner_cv,
    n_jobs=-1
)

# Outer loop: fair evaluation of the full tuning process.
nested_cv_scores = cross_val_score(
    nested_grid_search,
    X_train,
    y_train,
    scoring="f1",
    cv=outer_cv,
    n_jobs=-1
)

# Runtime.
nested_runtime = time.time() - nested_start_time

print("Nested CV completed.")
print("Nested CV F1 scores:", np.round(nested_cv_scores, 4))
print("Mean Nested CV F1:", round(nested_cv_scores.mean(), 4))
print("Std Nested CV F1:", round(nested_cv_scores.std(), 4))
print("Runtime seconds:", round(nested_runtime, 2))
```

<p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">
Grid Search and Random Search do not really learn from previous trials.

This can waste time because expensive model evaluations may be spent on weak parameter regions.

```python
# ============================================================
# CELL 76 — IMPORT BAYESIAN SEARCH TOOLS
# ============================================================

# BayesSearchCV performs Bayesian hyperparameter search.
from skopt import BayesSearchCV

# Integer and Real define the search range for hyperparameters.
from skopt.space import Integer, Real

print("Bayesian search tools imported successfully.")
```

```python
# ============================================================
# CELL 77 — BAYESIAN SEARCH WITH XGBOOST
# ============================================================

# Start timer because Bayesian tuning can take time.
bayes_start_time = time.time()

# Base XGBoost model.
# tree_method="hist" makes training faster on larger tabular data.

bayes_xgb_model = XGBClassifier(
    n_estimators=150,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric="auc",
    tree_method="hist",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Assignment search space.
# BayesSearchCV will search values between these ranges.
bayes_search_space = {
    "max_depth": Integer(3, 8),
    "learning_rate": Real(0.01, 0.30)
}

# Bayesian search with 30 iterations as required.
bayes_search = BayesSearchCV(
    estimator=bayes_xgb_model,
    search_spaces=bayes_search_space,
    n_iter=30,
    scoring="roc_auc",
    cv=3,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Train Bayesian search.
bayes_search.fit(X_santander_train, y_santander_train)

# Runtime.
bayes_runtime = time.time() - bayes_start_time

print("Bayesian Search completed.")
print("Best parameters:", bayes_search.best_params_)
print("Best CV AUC:", round(bayes_search.best_score_, 4))
print("Runtime seconds:", round(bayes_runtime, 2))
```

```python
# ============================================================
# CELL 82 — RANDOM SEARCH WITH XGBOOST
# ============================================================

# These distributions let RandomizedSearchCV sample values.
from scipy.stats import randint, uniform

# Start timer.
xgb_random_start_time = time.time()

# Base XGBoost model.
xgb_random_model = XGBClassifier(
    n_estimators=150,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric="auc",
    tree_method="hist",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Random Search space.
# max_depth: random integer from 3 to 8
# learning_rate: random value from 0.01 to 0.30
xgb_random_search_space = {
    "max_depth": randint(3, 9),
    "learning_rate": uniform(0.01, 0.29)
}

# RandomizedSearchCV tests 30 random combinations.
xgb_random_search = RandomizedSearchCV(
    estimator=xgb_random_model,
    param_distributions=xgb_random_search_space,
    n_iter=30,
    scoring="roc_auc",
    cv=3,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Train Random Search.
xgb_random_search.fit(X_santander_train, y_santander_train)

# Runtime.
xgb_random_runtime = time.time() - xgb_random_start_time

print("XGBoost Random Search completed.")
print("Best parameters:", xgb_random_search.best_params_)
print("Best CV AUC:", round(xgb_random_search.best_score_, 4))
print("Runtime seconds:", round(xgb_random_runtime, 2))
```

```python
# ============================================================
# BUILD KERASTUNER MODEL FUNCTION
# ============================================================

# This function builds one neural network.
# KerasTuner will call it many times with different choices.

def build_nn_model(hp):

    # Start a simple neural network.
    
    model = tf.keras.Sequential()

    # Input layer.
    # Santander has 200 numeric input features.
    
    model.add(tf.keras.layers.Input(shape=(X_nn_train.shape[1],)))

    # Tune how many hidden layers the network should have.
    
    number_of_layers = hp.Int(
        "num_layers",
        min_value=1,
        max_value=3
    )

    # Tune the activation function.
    
    activation_choice = hp.Choice(
        "activation",
        values=["relu", "tanh"]
    )

    # Add hidden layers based on the tuner choice.
    
    for layer_number in range(number_of_layers):

        # Tune how many neurons this layer should have.
        
        units = hp.Int(
            f"units_layer_{layer_number}",
            min_value=32,
            max_value=256,
            step=32
        )

        # Dense layer learns patterns from the previous layer.
        
        model.add(tf.keras.layers.Dense(
            units=units,
            activation=activation_choice
        ))

        # Tune dropout rate for this layer.
        dropout_rate = hp.Float(
            f"dropout_layer_{layer_number}",
            min_value=0.1,
            max_value=0.5,
            step=0.1
        )

        # Dropout helps reduce overfitting.
        model.add(tf.keras.layers.Dropout(rate=dropout_rate))

    # Output layer.
    # Sigmoid gives probability for class 1.
    model.add(tf.keras.layers.Dense(1, activation="sigmoid"))

    # Compile the model.
    # AUC is used because Santander is imbalanced.
    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=[tf.keras.metrics.AUC(name="auc")]
    )

    return model

print("KerasTuner model-building function created.")
```

<p style="margin:0; font-size:15px; line-height:1.7; color:#78350f;">
<b>Key takeaway:</b> This code does not train just one neural network. It runs a controlled experiment where 20 different neural network architectures are tested, compared using validation AUC, and saved for final model selection.
</p>

```python
# ============================================================
# RUN KERASTUNER RANDOMSEARCH
# ============================================================

# Early stopping stops training if validation AUC stops improving.
# This saves time and reduces overfitting risk.

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_auc",
    mode="max",
    patience=3,
    restore_best_weights=True
)

# RandomSearch will test 20 different neural network architectures.

nn_random_tuner = kt.RandomSearch(
    hypermodel=build_nn_model,
    objective=kt.Objective("val_auc", direction="max"),
    max_trials=20,
    executions_per_trial=1,
    directory=str(output_folder / "keras_tuner"),
    project_name="santander_nn_random_search",
    overwrite=True,
    seed=RANDOM_STATE
)

# Run the architecture search.

nn_random_start_time = time.time()

nn_random_tuner.search(
    X_nn_train,
    y_nn_train,
    validation_data=(X_nn_val, y_nn_val),
    epochs=10,
    batch_size=512,
    callbacks=[early_stop],
    verbose=1
)

nn_random_runtime = time.time() - nn_random_start_time

print("KerasTuner RandomSearch completed.")
print("Runtime seconds:", round(nn_random_runtime, 2))
```

```python
# ============================================================
# CELL 111B — BAYESIAN SEARCH VS RANDOM SEARCH CONVERGENCE
# ============================================================

# Convert CV results into DataFrames.
random_cv_results = pd.DataFrame(xgb_random_search.cv_results_)
bayes_cv_results = pd.DataFrame(bayes_search.cv_results_)

# Extract mean validation ROC-AUC scores from each trial.
random_scores = pd.to_numeric(random_cv_results["mean_test_score"])
bayes_scores = pd.to_numeric(bayes_cv_results["mean_test_score"])

# Best score found so far after each trial.
random_best_so_far = random_scores.cummax()
bayes_best_so_far = bayes_scores.cummax()

# Trial numbers.
trial_numbers = np.arange(1, len(random_best_so_far) + 1)

# Summary table for clean interpretation.
convergence_summary = pd.DataFrame({
    "Method": ["Random Search XGBoost", "Bayesian Search XGBoost"],
    "Best_CV_ROC_AUC": [random_best_so_far.iloc[-1], bayes_best_so_far.iloc[-1]],
    "Trial_Count": [len(random_best_so_far), len(bayes_best_so_far)]
})

convergence_summary["Best_CV_ROC_AUC"] = convergence_summary["Best_CV_ROC_AUC"].round(4)

display(convergence_summary)

# BI-ready chart.
fig, ax = plt.subplots(figsize=(12.5, 6.8), dpi=150)

ax.plot(
    trial_numbers,
    random_best_so_far,
    marker="o",
    linewidth=2.6,
    markersize=4,
    color="#64748B",
    label="Random Search"
)

ax.plot(
    trial_numbers,
    bayes_best_so_far,
    marker="o",
    linewidth=3.0,
    markersize=4,
    color="#1E3A8A",
    label="Bayesian Search"
)

# Highlight final values.
ax.scatter(
    trial_numbers[-1],
    random_best_so_far.iloc[-1],
    s=90,
    color="#64748B",
    zorder=5
)

ax.scatter(
    trial_numbers[-1],
    bayes_best_so_far.iloc[-1],
    s=110,
    color="#1E3A8A",
    zorder=5
)

ax.text(
    trial_numbers[-1] + 0.4,
    random_best_so_far.iloc[-1],
    f"{random_best_so_far.iloc[-1]:.4f}",
    va="center",
    fontsize=10,
    color="#334155"
)

ax.text(
    trial_numbers[-1] + 0.4,
    bayes_best_so_far.iloc[-1],
    f"{bayes_best_so_far.iloc[-1]:.4f}",
    va="center",
    fontsize=10,
    color="#172554",
    fontweight="bold"
)

# Titles.
fig.text(
    0.06,
    0.96,
    "XGBoost Hyperparameter Search Convergence",
    fontsize=18,
    fontweight="bold",
    color="#111827"
)

fig.text(
    0.06,
    0.925,
    "Best validation ROC-AUC found so far across 30 trials",
    fontsize=11,
    color="#4b5563"
)

# Axis formatting.
ax.set_xlabel("Trial number", fontsize=10.5, color="#374151")
ax.set_ylabel("Best CV ROC-AUC so far", fontsize=10.5, color="#374151")
ax.set_xlim(1, len(trial_numbers) + 2)

y_min = min(random_best_so_far.min(), bayes_best_so_far.min()) - 0.003
y_max = max(random_best_so_far.max(), bayes_best_so_far.max()) + 0.004
ax.set_ylim(y_min, y_max)

ax.grid(axis="y", alpha=0.18)
ax.set_axisbelow(True)

ax.legend(
    loc="lower right",
    frameon=False,
    fontsize=10
)

for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

fig.text(
    0.06,
    0.04,
    "Business note: Bayesian Search should improve search efficiency by learning from earlier trials. "
    "The final decision still depends on test-set performance, not CV score alone.",
    fontsize=10.2,
    color="#374151"
)

plt.tight_layout(rect=[0.04, 0.08, 0.98, 0.89])
plt.show()
```

33. KerasTuner Hyperband

<p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db; line-height:1.6;">
Hyperband is a smarter neural-network tuning method that avoids wasting full training time on weak architectures.
</p>
</div>

<br>

<div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#f9fafb; padding:15px 18px; border-radius:8px;">
<p style="margin:0; font-size:15px; line-height:1.7;">
Hyperband first trains many architectures for a small number of epochs. Then it keeps only the stronger candidates and gives them more training time. This idea is called <b>Successive Halving</b>.
</p>
</div>

</div>

```python
# ============================================================
# CELL 113B — RUN KERASTUNER HYPERBAND
# ============================================================

# Hyperband uses the same model-building function.
# It tests many architectures quickly, then continues only promising ones.
nn_hyperband_tuner = kt.Hyperband(
    hypermodel=build_nn_model,
    objective=kt.Objective("val_auc", direction="max"),
    max_epochs=15,
    factor=3,
    directory=str(output_folder / "keras_tuner"),
    project_name="santander_nn_hyperband",
    overwrite=True,
    seed=RANDOM_STATE
)

# Early stopping keeps training efficient.
hyperband_early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_auc",
    mode="max",
    patience=3,
    restore_best_weights=True
)

nn_hyperband_start_time = time.time()

nn_hyperband_tuner.search(
    X_nn_train,
    y_nn_train,
    validation_data=(X_nn_val, y_nn_val),
    epochs=15,
    batch_size=512,
    callbacks=[hyperband_early_stop],
    verbose=1
)

nn_hyperband_runtime = time.time() - nn_hyperband_start_time

print("KerasTuner Hyperband completed.")
print("Runtime seconds:", round(nn_hyperband_runtime, 2))
```
