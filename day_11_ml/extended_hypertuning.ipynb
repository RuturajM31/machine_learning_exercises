The training data is used for model fitting and cross-validation. The test data stays untouched until the final evaluation.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Additional validation split</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">A smaller train/validation split is created for optional optimisation methods such as Hyperopt and Optuna. This keeps those methods fast while still following a fair tuning pattern.</p></div></div>

```python
# CELL 4 — CREATE TRAIN-TEST SPLITS

# Classification split. Stratify keeps class balance similar in train and test sets.
X_cls_train, X_cls_test, y_cls_train, y_cls_test = train_test_split(
    X_classification,
    Y_classification,
    test_size=0.25,
    stratify=Y_classification,
    random_state=RANDOM_STATE
)

# Smaller train/validation split for optional optimisation libraries.
# This keeps Hyperopt, Optuna, and Halving fast and prevents local kernel crashes.
X_hpo_train, X_hpo_val, y_hpo_train, y_hpo_val = train_test_split(
    X_cls_train,
    y_cls_train,
    test_size=0.30,
    stratify=y_cls_train,
    random_state=RANDOM_STATE
)

# Regression split. Stratify is not used because the target is continuous.
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_regression,
    Y_regression,
    test_size=0.25,
    random_state=RANDOM_STATE
)

print("Classification train/test:", X_cls_train.shape, X_cls_test.shape)
print("HPO train/validation:", X_hpo_train.shape, X_hpo_val.shape)
print("Regression train/test:", X_reg_train.shape, X_reg_test.shape)
```

<h2 style="margin:0; font-size:23px; font-weight:800;">5. KNN Hyperparameter Tuning</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">KNN is sensitive to distance, scaling, and the number of neighbours.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">What KNN does</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">KNN predicts a new row by looking at the most similar rows in the training data. For classification, the nearby rows vote for the class. The model does not learn coefficients or trees; instead, it stores the training examples and uses distance at prediction time.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Why scaling is mandatory</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">KNN uses distance. If one feature has much larger numeric values than another, it can dominate the distance calculation. StandardScaler puts features on a comparable scale, making the neighbourhood search fairer.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #f59e0b; background:#fffbeb; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#78350f; font-size:19px;">Main tuning question</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">The key question is: how many neighbours should vote? A very small <code>k</code> may react too much to noise. A very large <code>k</code> may smooth away important patterns.</p></div></div>

<div style="font-family:Arial, sans-serif; background:#ffffff; color:#1f2937; line-height:1.65;">

<h2 style="margin:0; font-size:23px; font-weight:800;">KNN Concept Explanation</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Understand the main KNN hyperparameters before the code.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Important hyperparameters</h3><table style="width:100%; border-collapse:collapse; font-size:15px; line-height:1.6;"><tr><th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Hyperparameter</th><th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Simple meaning</th><th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Effect</th></tr><tr><td style="padding:9px; border-bottom:1px solid #e5e7eb;"><code>n_neighbors</code></td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Number of nearby rows used for voting</td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Controls flexibility vs smoothness</td></tr><tr><td style="padding:9px; border-bottom:1px solid #e5e7eb;"><code>weights='uniform'</code></td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Every neighbour has equal vote</td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Simple and stable</td></tr><tr><td style="padding:9px; border-bottom:1px solid #e5e7eb;"><code>weights='distance'</code></td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Closer neighbours vote more strongly</td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Can help when nearer examples are more reliable</td></tr><tr><td style="padding:9px; ">Scaling step</td><td style="padding:9px; ">Standardises numeric features</td><td style="padding:9px; ">Prevents large-scale variables from dominating distance</td></tr></table></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">What this section will do</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">GridSearchCV tests multiple KNN settings using cross-validation and selects the setup with the strongest F1-score. F1-score is useful because it balances precision and recall for the positive class.</p></div></div>

```python
# CELL 5 — KNN GRID SEARCH

# Pipeline keeps scaling and KNN together.
knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])

# Small readable search space.
knn_grid = {
    "model__n_neighbors": [3, 5, 7],
    "model__weights": ["uniform", "distance"]
}

# Tune KNN using F1-score.
knn_search = GridSearchCV(
    knn_pipeline,
    knn_grid,
    scoring="f1",
    cv=3,
    n_jobs=1
)

knn_search.fit(X_cls_train, y_cls_train)

print("Best KNN parameters:", knn_search.best_params_)
print("Best CV F1:", round(knn_search.best_score_, 4))
```

<h2 style="margin:0; font-size:23px; font-weight:800;">5.1 Evaluate Tuned KNN</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Check the tuned KNN model on the untouched classification test set.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Why this step is needed</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Cross-validation chooses hyperparameters, but final evaluation must happen on the test set. <h2 style="margin:0; font-size:23px; font-weight:800;">6. Logistic Regression Hyperparameter Tuning</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Logistic Regression is an interpretable classification baseline.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">What Logistic Regression does</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Logistic Regression estimates the probability that a row belongs to class 1. It is not a deep model; it learns weighted relationships between the input features and the target probability.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Why it is useful</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">It is fast, stable, and easier to explain than many complex models. In business settings, this matters because stakeholders often want to understand why a prediction changes.</p></div></div>

<div style="font-family:Arial, sans-serif; background:#ffffff; color:#1f2937; line-height:1.65;">
<div style="background:#111827; color:white; padding:16px 20px; border-radius:8px;">
<h2 style="margin:0; font-size:23px; font-weight:800;">Logistic Regression Concept Explanation</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Understand regularisation before tuning.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Main hyperparameters</h3><table style="width:100%; border-collapse:collapse; font-size:15px; line-height:1.6;"><tr><th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Hyperparameter</th><th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Meaning</th><th style="text-align:left; padding:9px; border-bottom:2px solid #d1d5db;">Effect</th></tr><tr><td style="padding:9px; border-bottom:1px solid #e5e7eb;"><code>C</code></td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Inverse regularisation strength</td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Smaller C means stronger regularisation</td></tr><tr><td style="padding:9px; border-bottom:1px solid #e5e7eb;"><code>penalty='l2'</code></td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Shrinks coefficients smoothly</td><td style="padding:9px; border-bottom:1px solid #e5e7eb;">Reduces overfitting</td></tr><tr><td style="padding:9px; "><code>solver</code></td><td style="padding:9px; ">Optimisation algorithm</td><td style="padding:9px; ">Controls how the model is fitted</td></tr></table></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #f59e0b; background:#fffbeb; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#78350f; font-size:19px;">Simple intuition</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Regularisation is like asking the model not to overreact. A strongly regularised model keeps coefficients smaller and can generalise better when data has noise.</p></div></div>

```python
# CELL 7 — LOGISTIC REGRESSION GRID SEARCH

# Pipeline scales features and trains Logistic Regression.
logreg_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=1000,
        solver="liblinear",
        random_state=RANDOM_STATE
    ))
])

# Tune regularisation strength and penalty type.
logreg_grid = {
    "model__C": [0.01, 0.1, 1, 10],
    "model__penalty": ["l1", "l2"]
}

logreg_search = GridSearchCV(
    logreg_pipeline,
    logreg_grid,
    scoring="roc_auc",
    cv=3,
    n_jobs=1
)

logreg_search.fit(X_cls_train, y_cls_train)

print("Best Logistic Regression parameters:", logreg_search.best_params_)
print("Best CV ROC-AUC:", round(logreg_search.best_score_, 4))
```

<h2 style="margin:0; font-size:23px; font-weight:800;">7. Ridge Regression Hyperparameter Tuning</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Ridge covers the regression side of hyperparameter tuning.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">What Ridge Regression does</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Ridge Regression predicts a numeric target. It is Linear Regression with an added regularisation penalty that keeps coefficients smaller and more stable.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Why it is included</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Hyperparameter tuning is not only for classification. Regression models also have hyperparameters, and their performance must be measured with regression metrics such as RMSE, MAE, and R².</p></div></div>

<h2 style="margin:0; font-size:23px; font-weight:800;">Ridge Regression Concept Explanation</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Understand alpha and regression metrics before the code.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Main hyperparameter</h3><table style="width:100%;
```python
# CELL 9 — RIDGE REGRESSION GRID SEARCH

# Pipeline scales features and trains Ridge Regression.
ridge_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", Ridge())
])

# Tune the regularisation strength.
ridge_grid = {
    "model__alpha": [0.01, 0.1, 1, 10, 100]
}

ridge_search = GridSearchCV(
    ridge_pipeline,
    ridge_grid,
    scoring="neg_root_mean_squared_error",
    cv=3,
    n_jobs=1
)

ridge_search.fit(X_reg_train, y_reg_train)

print("Best Ridge parameters:", ridge_search.best_params_)
print("Best CV RMSE:", round(-ridge_search.best_score_, 4))
```

Here we evaluate numeric prediction error rather than class probabilities.</p></div></div>

```python
# CELL 10 — EVALUATE TUNED RIDGE REGRESSION

# Predict numeric target values.
ridge_prediction = ridge_search.best_estimator_.predict(X_reg_test)

# Calculate regression metrics.
ridge_rmse = mean_squared_error(y_reg_test, ridge_prediction) ** 0.5
ridge_mae = mean_absolute_error(y_reg_test, ridge_prediction)
ridge_r2 = r2_score(y_reg_test, ridge_prediction)

ridge_result = {
    "Model": "Ridge Regression GridSearch",
    "RMSE": ridge_rmse,
    "MAE": ridge_mae,
    "R2": ridge_r2
}

pd.DataFrame([ridge_result]).round(4)
```

<h2 style="margin:0; font-size:23px; font-weight:800;">8. CASH and AutoML Practical Workflow</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">CASH means choosing both the algorithm and its hyperparameters.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Simple idea</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Normal hyperparameter tuning asks: which settings are best for this one model? CASH asks a bigger question: which model family and which settings are best together?</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Why this matters</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">In real projects, selecting the wrong model family can limit performance before tuning even starts. They can search over preprocessing choices, algorithms, hyperparameters, and sometimes ensemble combinations.</p></div></div>

```python
# CELL 11 — CASH MODEL FAMILY COMPARISON

# Compare tuned/standard model families on the same training data.
cash_candidates = {
    "KNN": knn_search.best_estimator_,
    "Logistic Regression": logreg_search.best_estimator_,
    "Random Forest": RandomForestClassifier(
        n_estimators=80,
        max_depth=5,
        random_state=RANDOM_STATE,
        n_jobs=1
    ),
    "SVC": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(C=1, kernel="rbf", probability=True, random_state=RANDOM_STATE))
    ])
}

cash_rows = []

# Cross-validate each model family using ROC-AUC.
for model_name, model_object in cash_candidates.items():
    cv_scores = cross_val_score(
        model_object,
        X_cls_train,
        y_cls_train,
        scoring="roc_auc",
        cv=3,
        n_jobs=1
    )
    cash_rows.append({"Model": model_name, "CV_ROC_AUC": cv_scores.mean()})

cash_results = pd.DataFrame(cash_rows).sort_values("CV_ROC_AUC", ascending=False)
cash_results.round(4)
```

Deeper Bayesian Optimisation Theory</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">This expands the Bayesian optimisation topic beyond the main MVD notebook.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Why Bayesian optimisation exists</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Grid Search can waste time checking every combination. Random Search samples blindly. Bayesian optimisation tries to be smarter by using previous results to decide what to test next.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Main idea</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">When model training is expensive, each trial matters. Bayesian optimisation builds a probabilistic model of the relationship between hyperparameters and validation score, then uses that model to choose promising next trials.</p></div></div>

```python
# CELL 13 — IMPORT HYPEROPT SAFELY

# Hyperopt is optional in many local environments.
try:
    from hyperopt import fmin, tpe, hp, Trials, STATUS_OK
    HYPEROPT_AVAILABLE = True
    print("Hyperopt available: True")
except Exception as error:
    HYPEROPT_AVAILABLE = False
    print("Hyperopt available: False")
    print("Reason:", str(error)[:120])
```

<h2 style="margin:0; font-size:23px; font-weight:800;">10.1 Define Hyperopt Objective and Search Space</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Prepare the optimisation logic before running the search.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">What the objective does</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">The objective function receives one hyperparameter configuration, trains a Random Forest on the HPO training split, predicts probabilities on the validation split, calculates ROC-AUC, and returns validation loss.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">Why this is separated</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">Keeping the objective function in its own cell makes the code easier to understand and avoids one long complicated block.</p></div></div>

```python
# CELL 14A — DEFINE HYPEROPT OBJECTIVE AND SEARCH SPACE

# This objective trains one Random Forest setting and returns validation loss.
def rf_validation_loss(n_estimators, max_depth, min_samples_leaf):
    model = RandomForestClassifier(
        n_estimators=int(n_estimators),
        max_depth=int(max_depth),
        min_samples_leaf=int(min_samples_leaf),
        random_state=RANDOM_STATE,
        n_jobs=1
    )
    model.fit(X_hpo_train, y_hpo_train)
    probability = model.predict_proba(X_hpo_val)[:, 1]
    return -roc_auc_score(y_hpo_val, probability)

# Small search values keep the notebook fast and stable.
hyperopt_candidate_grid = [
    {"n_estimators": 25, "max_depth": 2, "min_samples_leaf": 2},
    {"n_estimators": 50, "max_depth": 3, "min_samples_leaf": 2},
    {"n_estimators": 75, "max_depth": 4, "min_samples_leaf": 2},
]

print("Hyperopt objective and search values are ready.")
```

<div style="background:#111827; color:white; padding:16px 20px; border-radius:8px;">
<h2 style="margin:0; font-size:23px; font-weight:800;">Hyperopt Objective Definition Conclusion</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">The objective function is ready.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #16a34a; background:#f0fdf4; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#166534; font-size:19px;">What is ready now</h3><ul style="margin:0; padding-left:22px; font-size:15px; line-height:1.75;"><li>A validation-loss function based on ROC-AUC.</li><li>A small set of Random Forest hyperparameter values.</li><li>A structure that Hyperopt or the fallback search can use safely.</li></ul></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">Interpretation</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">This cell does not yet choose the best model. It defines the scoring rule that the next optimisation cell will use.</p></div></div>

<div style="font-family:Arial, sans-serif; background:#ffffff; color:#1f2937; line-height:1.65;">
<div style="background:#111827; color:white; padding:16px 20px; border-radius:8px;">
<h2 style="margin:0; font-size:23px; font-weight:800;">10.2 Run Hyperopt Optimisation</h2><p style="margin:7px 0 0 0; font-size:14.5px; color:#d1d5db;">Search for the Random Forest settings with the lowest validation loss.</p></div><br><div style="border:1px solid #e5e7eb; border-left:6px solid #2563eb; background:#eff6ff; padding:15px 18px; border-radius:8px;"><h3 style="margin:0 0 10px 0; color:#1e40af; font-size:19px;">What happens here</h3><p style="margin:0 0 10px 0; font-size:15px; line-height:1.7;">If Hyperopt is available, the TPE algorithm tests different Random Forest settings. If Hyperopt is not available, the notebook uses a safe local fallback over the same compact candidate list.</p></div></div>

```python
# CELL 14B — RUN HYPEROPT OR SAFE FALLBACK

if HYPEROPT_AVAILABLE:
    # Real Hyperopt search space.
    hyperopt_space = {
        "n_estimators": hp.quniform("n_estimators", 25, 75, 25),
        "max_depth": hp.quniform("max_depth", 2, 5, 1),
        "min_samples_leaf": hp.quniform("min_samples_leaf", 1, 3, 1)
    }

    # Hyperopt minimises loss, so the objective returns negative ROC-AUC.
    def hyperopt_objective(params):
        loss = rf_validation_loss(params["n_estimators"], params["max_depth"], params["min_samples_leaf"])
        return {"loss": loss, "status": STATUS_OK}

    hyperopt_trials = Trials()
    hyperopt_best = fmin(hyperopt_objective, hyperopt_space, algo=tpe.suggest, max_evals=6, trials=hyperopt_trials, rstate=np.random.default_rng(RANDOM_STATE), show_progressbar=False)
else:
    # Safe fallback: test the same idea manually without external dependency.
    fallback_rows = []
    for params in hyperopt_candidate_grid:
        loss = rf_validation_loss(**params)
        fallback_rows.append({**params, "loss": loss})
    hyperopt_best = pd.DataFrame(fallback_rows).sort_values("loss").iloc[0].to_dict()

print("Best Hyperopt-style settings:")
print(hyperopt_best)
```

```python
# CELL 20 — RUN SUCCESSIVE HALVING OR SAFE FALLBACK

if HALVING_AVAILABLE:
    halving_search = HalvingRandomSearchCV(
        estimator=RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=1),
        param_distributions={
            "max_depth": randint(2, 6),
            "min_samples_leaf": randint(1, 4)
        },
        scoring="roc_auc",
        cv=2,
        factor=2,
        resource="n_estimators",
        min_resources=10,
        max_resources=40,
        random_state=RANDOM_STATE,
        n_jobs=1
    )
    halving_search.fit(X_hpo_train, y_hpo_train)
    halving_best_model = halving_search.best_estimator_
else:
    # Fallback uses the Optuna-selected model.
    halving_best_model = optuna_model

print("Successive Halving / fallback model ready.")
```
