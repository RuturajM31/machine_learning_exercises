<h3 style="margin:0 0 8px 0; color:#1e3a8a; font-size:19px;">Step 5 — Regression extra topic: predicting numeric values</h3>
<p style="margin:8px 0;"><strong>Simple idea:</strong> regression predicts a number, not a category. For example, it can estimate price, demand, delivery time, or cost.</p>
<p style="margin:8px 0;">The uploaded PyCaret workflow scripts use a regression-style demonstration to load data, run setup, compare models, finalize a champion, predict, and evaluate metrics. Here we use a stable local scikit-learn dataset to show the same regression idea without relying on internet downloads.</p>
</div>

```python

# Load the diabetes regression dataset from sklearn.
diabetes = load_diabetes(as_frame=True)

# Store the feature table.
X_reg = diabetes.data

# Store the numeric target.
y_reg = diabetes.target

# Split data into train and test parts.
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.25, random_state=RANDOM_STATE
)

# Create a simple baseline regression model.
reg_model = LinearRegression()

# Fit the regression model on training data.
reg_model.fit(X_train_reg, y_train_reg)

# Predict target values on the test data.
reg_predictions = reg_model.predict(X_test_reg)

# Calculate mean absolute error.
reg_mae = mean_absolute_error(y_test_reg, reg_predictions)

# Calculate R-squared.
reg_r2 = r2_score(y_test_reg, reg_predictions)

# Create a compact metrics table.
reg_metrics = pd.DataFrame({
    "Metric": ["MAE", "R2"],
    "Value": [reg_mae, reg_r2]
})

# Display the metrics.
display(reg_metrics.round(4))
```

<h3 style="margin:0 0 8px 0; color:#14532d; font-size:19px;">Step 5 conclusion</h3>

<h3 style="margin:0 0 8px 0; color:#1e3a8a; font-size:19px;">Step 11 — Time series extra topic: forecasting ordered data</h3>
<p style="margin:8px 0;"><strong>Simple idea:</strong> time series forecasting predicts future values from past ordered values. Unlike normal tabular ML, the order of time matters.</p>
<p style="margin:8px 0;">Examples: monthly sales, weekly shipments, daily website traffic, or hourly production volume.</p>
</div>

```python

# Create a monthly date range.
dates = pd.date_range(start="2024-01-01", periods=36, freq="MS")

# Create a stable trend component.
trend = np.linspace(100, 165, len(dates))

# Create a seasonal wave component.
seasonality = 12 * np.sin(np.arange(len(dates)) * 2 * np.pi / 12)

# Create stable random noise.
rng = np.random.default_rng(RANDOM_STATE)
noise = rng.normal(0, 4, len(dates))

# Combine components into a synthetic demand series.
demand = trend + seasonality + noise

# Store the time series in a DataFrame.
ts_df = pd.DataFrame({"Month": dates, "Demand": demand})

# Create a simple lag feature from the previous month.
ts_df["Lag_1"] = ts_df["Demand"].shift(1)

# Remove the first row because its lag is missing.
ts_model_df = ts_df.dropna().copy()

# Display the first rows.
display(ts_model_df.head())
```

```python
# Import text tools for a small NLP mini-demo.
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Create a tiny text dataset.
text_samples = [
    "delivery arrived on time",
    "shipment delayed again",
    "excellent service and fast delivery",
    "late package and bad communication",
    "smooth logistics process",
    "delivery failed and customer complained"
]

# Create labels: 1 means positive, 0 means negative.
text_labels = np.array([1, 0, 1, 0, 1, 0])

# Convert text into word-count features.
vectorizer = CountVectorizer()
text_features = vectorizer.fit_transform(text_samples)

# Train a simple Naive Bayes text classifier.
text_model = MultinomialNB()
text_model.fit(text_features, text_labels)

# Create and transform two example messages.
new_messages = ["fast delivery and smooth service", "shipment delayed and failed"]
new_features = vectorizer.transform(new_messages)

# Predict sentiment-style labels.
text_predictions = text_model.predict(new_features)

# Build a readable result table.
text_result = pd.DataFrame({
    "Message": new_messages,
    "Predicted_Label": text_predictions,
    "Meaning": np.where(text_predictions == 1, "Positive", "Negative")
})

# Display the result.
display(text_result)
```
