<div class="ld10">
<h2>4. Random Forest recap, OOB and bias-variance</h2>
<p>Random Forest was covered in the previous topic, but it is important here because XGBoost is often compared against it.</p>
<p><b>Random Forest idea:</b> train many different trees and combine their votes or averages.</p>
<p><b>OOB validation:</b> Random Forest can use rows left out of each bootstrap sample as a built-in validation check.</p>
<p><b>Bias-variance recap:</b></p>
<ul>
<li>High bias means the model is too simple.</li>
<li>High variance means the model is too sensitive to training data.</li>
<li>Random Forest usually reduces variance.</li>
<li>Boosting often reduces bias, but it must be controlled to avoid overfitting.</li>
</ul>
</div>

<div class="ld10">
<h2>5. Weak learners and decision stumps</h2>
<p>Boosting often uses weak learners. A weak learner is a simple model that is only slightly better than guessing.</p>
<p>A common weak learner is a small decision tree:</p>
<ul>
<li>depth 1: decision stump</li>
<li>depth 2 or 3: small tree</li>
<li>many small trees together: strong boosted model</li>
</ul>
<p>This is why boosting can turn many simple models into one powerful model.</p>
</div>

<div class="ld10">
<h2>6. AdaBoost</h2>
<p>AdaBoost is an older boosting method.</p>
<p>Its main idea is simple:</p>
<ol>
<li>Train a small model.</li>
<li>Find the rows it predicted wrongly.</li>
<li>Give those wrong rows more weight.</li>
<li>Train the next model with more focus on those hard rows.</li>
</ol>
</div>

<h2>7. Gradient Boosting</h2>
<p>Gradient Boosting improves the general boosting idea.</p>
<p>Instead of only increasing row weights, it trains each new model to reduce the current model error.</p>
<p>For regression, this is easy to understand:</p>
<ul>
<li>first model predicts a rough value</li>
<li>next model learns the remaining error</li>
<li>another model learns the remaining error again</li>
<li>the final model is the sum of many small corrections</li>
</ul>
</div>

<h2>8. Forward stagewise additive modelling</h2>
<p>This is the formal theory behind Gradient Boosting.</p>
<p>In simple words:</p>
<ul>
<li><b>Forward</b> means we move step by step.</li>
<li><b>Stagewise</b> means we add one model at a time.</li>
<li><b>Additive</b> means each new model is added to the existing model.</li>
</ul>
<p>So the final model looks like this:</p>
<p><code>Final prediction = model 1 + model 2 + model 3 + ...</code></p>
<p>Each tree does not solve the full problem. It only adds a correction.</p>
</div>

<div class="ld10">
<h2>9. Loss functions and pseudo-residuals</h2>
<p>A loss function tells the model how bad its prediction is.</p>
<p>Examples:</p>
<ul>
<li>Classification: wrong class prediction creates classification loss.</li>
<li>Regression: wrong price prediction creates RMSE or squared error loss.</li>
</ul>
<p>A residual means:</p>
<p><code>actual value - predicted value</code></p>
<p>A pseudo-residual is the correction direction used by Gradient Boosting when the loss function is more complex.</p>
<p>Simple meaning:</p>
<p><b>The next tree learns where the current model should move to reduce error.</b></p>
</div>

<div class="ld10">
<h2>10. Learning rate, number of trees and early stopping</h2>
<p>These are the most important boosting controls.</p>
<table class="ld10-table"><thead><tr><th>Parameter</th><th>Simple meaning</th></tr></thead><tbody><tr><td><code>learning_rate</code></td><td>How strong each new tree correction is</td></tr><tr><td><code>n_estimators</code></td><td>How many trees are added</td></tr><tr><td><code>max_depth</code></td><td>How complex each tree can become</td></tr><tr><td><code>early_stopping_rounds</code></td><td>Stop when validation score stops improving</td></tr></tbody></table>
<p>A small learning rate is safer but usually needs more trees. Early stopping prevents the model from training longer than useful.</p>

```python
# ============================================================
# DATASET LOADING HELPERS
# ============================================================
# These helpers load Kaggle files while keeping the notebook clean.

def unzip_first_zip(folder):
    # Unzip the first zip file in a folder if one exists.
    folder = Path(folder)
    zip_files = list(folder.glob("*.zip"))

    for zip_path in zip_files:
        with zipfile.ZipFile(zip_path, "r") as zip_ref:
            zip_ref.extractall(folder)


def try_kaggle_dataset_download(dataset_slug, output_folder):
    # Try to download a Kaggle dataset using kagglehub or Kaggle CLI.
    # This is used for the Bank Marketing dataset.
    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)

    # First try kagglehub.
    try:
        import kagglehub
        downloaded_path = Path(kagglehub.dataset_download(dataset_slug))
        print("KaggleHub download completed.")
        return downloaded_path
    except Exception as error:
        print("KaggleHub dataset download did not work.")
        print(f"Reason: {error}")

    # Then try Kaggle CLI.
    try:
        command = [
            "kaggle", "datasets", "download",
            "-d", dataset_slug,
            "-p", str(output_folder)
        ]
        subprocess.run(command, check=True)
        unzip_first_zip(output_folder)
        print("Kaggle CLI dataset download completed.")
        return output_folder
    except Exception as error:
        print("Kaggle CLI dataset download did not work.")
        print(f"Reason: {error}")

    return None


def try_kaggle_competition_download(competition_slug, output_folder):
    # Try to download a Kaggle competition dataset using Kaggle CLI.
    # This is used for the House Prices dataset.
    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)

    try:
        command = [
            "kaggle", "competitions", "download",
            "-c", competition_slug,
            "-p", str(output_folder)
        ]
        subprocess.run(command, check=True)
        unzip_first_zip(output_folder)
        print("Kaggle competition download completed.")
        return output_folder
    except Exception as error:
        print("Kaggle competition download did not work.")
        print(f"Reason: {error}")
        return None


def find_csv_file(folder, keywords):
    # Find the first CSV file whose name contains one of the expected keywords.
    folder = Path(folder)

    if not folder.exists():
        return None

    # Search inside subfolders too.
    csv_files = list(folder.rglob("*.csv"))

    # Prefer files that contain expected keywords.
    for csv_file in csv_files:
        file_name = csv_file.name.lower()
        if any(keyword.lower() in file_name for keyword in keywords):
            return csv_file

    # If no keyword match is found, return the first CSV file.
    return csv_files[0] if csv_files else None
```

<h2>13. Encoding and train/test split</h2>
<p>XGBoost needs numeric input. For this assignment-style classification part, categorical columns are encoded with <code>LabelEncoder</code>.</p>
<p>This is simple and fast, but we must remember that label numbers do not mean natural order for categories.</p>

```python
# ============================================================
# BANK MARKETING ENCODING
# ============================================================
# This cell separates inputs and target, then encodes categorical columns.

# Copy the original data so the raw data stays unchanged.
bank_model_data = bank_data.copy()

# Separate input columns from target column.
# X_bank contains customer and campaign features.
# y_bank_raw contains the deposit decision.
X_bank = bank_model_data.drop(columns=[bank_target_column])
y_bank_raw = bank_model_data[bank_target_column]

# Encode the target column.
# Example: no/yes becomes 0/1.
bank_target_encoder = LabelEncoder()
y_bank = bank_target_encoder.fit_transform(y_bank_raw)

# Identify categorical columns.
bank_categorical_columns = X_bank.select_dtypes(include=["object", "category"]).columns.tolist()

# Identify numeric columns.
bank_numeric_columns = X_bank.select_dtypes(exclude=["object", "category"]).columns.tolist()

# Encode each categorical column with its own LabelEncoder.
bank_feature_encoders = {}

for column in bank_categorical_columns:
    # Create a copy-safe string version.
    # Missing values are converted into the word "Missing".
    X_bank[column] = X_bank[column].astype(str).fillna("Missing")

    # Create one encoder per categorical column.
    encoder = LabelEncoder()

    # Transform the text categories into numbers.
    X_bank[column] = encoder.fit_transform(X_bank[column])

    # Store the encoder in case we need to inspect mappings later.
    bank_feature_encoders[column] = encoder

# Fill any numeric missing values with median.
for column in bank_numeric_columns:
    X_bank[column] = X_bank[column].fillna(X_bank[column].median())

print(f"Categorical columns encoded: {len(bank_categorical_columns)}")
print(f"Numeric columns used: {len(bank_numeric_columns)}")
print(f"Target classes: {list(bank_target_encoder.classes_)}")

display(X_bank.head())
```

<h2>17. Preprocessing for XGBoost and LightGBM</h2>
<p>XGBoost and LightGBM need numeric model-ready input.</p>
<p>The preprocessing pipeline does this:</p>
<ul>
<li>fills missing numeric values with the median</li>
<li>fills missing categorical values with <code>&quot;Missing&quot;</code></li>
<li>one-hot encodes categorical columns</li>
<li>passes numeric columns through as numeric values</li>
</ul>
</div>

```python
# ============================================================
# PREPROCESSING PIPELINE FOR XGBOOST AND LIGHTGBM
# ============================================================
# This pipeline turns raw House Prices features into numeric model-ready features.

# Numeric pipeline:
# Missing numeric values are replaced with the median.
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

# Categorical pipeline:
# Missing categories are replaced with "Missing".
# OneHotEncoder turns categories into numeric columns.
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="constant", fill_value="Missing")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

# ColumnTransformer applies the correct pipeline to each column type.
house_preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, house_numeric_columns),
        ("cat", categorical_transformer, house_categorical_columns)
    ],
    remainder="drop"
)

print("Preprocessing pipeline created.")
```

<div class="ld10-conclusion ">
  <h3>Output interpretation — Preprocessing pipeline</h3>
  
  <p>The preprocessing pipeline is ready. Numeric features are imputed and categorical features are encoded before XGBoost and LightGBM use them.</p><p>The pipeline approach is important because it keeps data preparation consistent across models and reduces the risk of leakage.</p>
</div>

<h2>19. XGBoost tuning with GridSearchCV</h2>
<p>GridSearchCV tests a fixed list of parameter combinations.</p>
<p>In fast mode, the grid is smaller so the notebook runs quickly. In full mode, the wider required grid can be used.</p>
</div>

```python
# ============================================================
# XGBOOST GRIDSEARCHCV
# ============================================================
# This cell tunes XGBoost hyperparameters.

# Full required grid.
xgb_required_grid = {
    "model__max_depth": [3, 5, 7],
    "model__learning_rate": [0.01, 0.1, 0.3],
    "model__n_estimators": [100, 300]
}

# Faster grid for quick checking.
xgb_fast_grid = {
    "model__max_depth": [3, 5],
    "model__learning_rate": [0.05, 0.1],
    "model__n_estimators": [100]
}

# Select the grid based on run mode.
xgb_param_grid = xgb_fast_grid if FAST_MODE else xgb_required_grid

# Create a fresh model for tuning.
xgb_regressor_for_grid = XGBRegressor(
    objective="reg:squarederror",
    subsample=0.85,
    colsample_bytree=0.85,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

# Create pipeline for grid search.
xgb_grid_pipeline = Pipeline(steps=[
    ("preprocessor", house_preprocessor),
    ("model", xgb_regressor_for_grid)
])

# GridSearchCV uses negative RMSE because scikit-learn maximizes scores.
xgb_grid_search = GridSearchCV(
    estimator=xgb_grid_pipeline,
    param_grid=xgb_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=3,
    n_jobs=-1,
    verbose=1
)

# Train the grid search.
xgb_grid_search.fit(X_house_train, y_house_train)

# Store best model and prediction.
best_xgb_model = xgb_grid_search.best_estimator_
xgb_tuned_pred = best_xgb_model.predict(X_house_valid)
xgb_tuned_rmse = calculate_rmse(y_house_valid, xgb_tuned_pred)

print("Best XGBoost parameters:")
print(xgb_grid_search.best_params_)
print(f"Best CV RMSE: ${-xgb_grid_search.best_score_:,.2f}")
print(f"Validation RMSE: ${xgb_tuned_rmse:,.2f}")
```

<h2>20. CatBoost regression</h2>
<p>CatBoost is useful because it can work with categorical columns directly.</p>
<p>Instead of manually one-hot encoding every category, we tell CatBoost which columns are categorical.</p>
</div>

```python
# ============================================================
# CATBOOST DATA PREPARATION
# ============================================================
# CatBoost can use raw categorical columns, but missing values must be handled clearly.

# Copy raw train and validation data.
X_cat_train = X_house_train.copy()
X_cat_valid = X_house_valid.copy()

# Fill numeric missing values with median.
for column in house_numeric_columns:
    median_value = X_cat_train[column].median()
    X_cat_train[column] = X_cat_train[column].fillna(median_value)
    X_cat_valid[column] = X_cat_valid[column].fillna(median_value)

# Fill categorical missing values with the word "Missing".
# Convert categories to strings because CatBoost expects categorical values clearly.
for column in house_categorical_columns:
    X_cat_train[column] = X_cat_train[column].astype(str).fillna("Missing")
    X_cat_valid[column] = X_cat_valid[column].astype(str).fillna("Missing")

# CatBoost needs categorical feature indices, not names.
cat_feature_indices = [
    X_cat_train.columns.get_loc(column)
    for column in house_categorical_columns
]

print(f"CatBoost categorical features: {len(cat_feature_indices)}")
```

  <p>CatBoost receives the raw categorical feature positions directly. This is the main reason CatBoost is useful: it can learn from category columns without forcing us to manually one-hot encode every category.</p><p>This is valuable for real business datasets with many columns such as region, product category, channel, customer type or house style.</p>

```python
# ============================================================
# PREPARE PROCESSED DATA FOR LIGHTGBM AND OPTUNA
# ============================================================
# We fit the preprocessor on training data only.
# Then we transform train and validation data.

# Fit the preprocessor.
house_preprocessor.fit(X_house_train)

# Transform raw data into numeric matrices.
X_house_train_processed = house_preprocessor.transform(X_house_train)
X_house_valid_processed = house_preprocessor.transform(X_house_valid)

print("Processed data ready.")
print(f"Train processed shape: {X_house_train_processed.shape}")
print(f"Valid processed shape: {X_house_valid_processed.shape}")
```

```python
# ============================================================
# LIGHTGBM BASELINE MODEL
# ============================================================
# This gives us a LightGBM starting point before Optuna tuning.

lgbm_baseline = LGBMRegressor(
    n_estimators=FAST_N_ESTIMATORS if FAST_MODE else FULL_N_ESTIMATORS,
    learning_rate=0.06,
    num_leaves=40,
    min_child_samples=20,
    random_state=RANDOM_STATE,
    n_jobs=-1,
    verbose=-1
)

# Train LightGBM baseline.
lgbm_baseline.fit(X_house_train_processed, y_house_train)

# Predict validation prices.
lgbm_baseline_pred = lgbm_baseline.predict(X_house_valid_processed)

# Calculate RMSE.
lgbm_baseline_rmse = calculate_rmse(y_house_valid, lgbm_baseline_pred)

print(f"LightGBM baseline RMSE: ${lgbm_baseline_rmse:,.2f}")
```

  <p>The LightGBM baseline is weaker than XGBoost and CatBoost in this fast run. This does not mean LightGBM is a bad model; it means the initial settings are not strong enough for this dataset.</p><p>The next Optuna step is therefore useful because LightGBM often improves significantly when key parameters are tuned.</p>
</div>

</div>

```python
# ============================================================
# OPTUNA OBJECTIVE FUNCTION FOR LIGHTGBM
# ============================================================
# This function tells Optuna how to test one LightGBM configuration.

def lightgbm_objective(trial):
    """Train one LightGBM model and return validation RMSE."""

    # Ask Optuna to suggest parameter values.
    params = {
        "num_leaves": trial.suggest_int("num_leaves", 20, 300),
        "learning_rate": trial.suggest_float("learning_rate", 0.001, 0.3, log=True),
        "min_child_samples": trial.suggest_int("min_child_samples", 5, 100),

        # Support parameters to keep the model stable.
        "n_estimators": 700 if FAST_MODE else 1500,
        "subsample": 0.85,
        "colsample_bytree": 0.85,
        "random_state": RANDOM_STATE,
        "n_jobs": -1,
        "verbose": -1
    }

    # Create model with suggested parameters.
    model = LGBMRegressor(**params)

    # Train model.
    model.fit(
        X_house_train_processed,
        y_house_train,
        eval_set=[(X_house_valid_processed, y_house_valid)],
        eval_metric="rmse"
    )

    # Predict validation values.
    predictions = model.predict(X_house_valid_processed)

    # Return RMSE because Optuna will minimise it.
    rmse = calculate_rmse(y_house_valid, predictions)
    return rmse


# Create the Optuna study.
# direction="minimize" because lower RMSE is better.
optuna.logging.set_verbosity(optuna.logging.WARNING)

lgbm_study = optuna.create_study(
    direction="minimize",
    study_name="lightgbm_house_prices_tuning"
)

# Run the study.
lgbm_study.optimize(lightgbm_objective, n_trials=OPTUNA_TRIALS)

print(f"Optuna trials completed: {len(lgbm_study.trials)}")
print(f"Best LightGBM RMSE: ${lgbm_study.best_value:,.2f}")
print("Best parameters:")
print(lgbm_study.best_params)
```

<div class="ld10">

<div class="ld10-conclusion ld10-success">
  <h3>Output interpretation — Optuna LightGBM tuning</h3>
  <div class="ld10-metric-row"><span class="ld10-metric ">Trials: <b>10</b></span><span class="ld10-metric ">Best RMSE: <b>$26,858.08</b></span><span class="ld10-metric ">num_leaves: <b>237</b></span><span class="ld10-metric ">learning_rate: <b>0.0129</b></span><span class="ld10-metric ">min_child_samples: <b>11</b></span></div>
  <p>Optuna improved LightGBM from the baseline RMSE of $28,990.67 to $26,858.08. That is a meaningful improvement, showing that LightGBM benefits from tuning.</p><p>Because FAST_MODE used 10 trials instead of 50, this is a fast proof of the tuning process. A full run may find a better setting.</p>
</div>

```python
# ============================================================
# TRAIN BEST LIGHTGBM MODEL FROM OPTUNA
# ============================================================
# This cell trains the final tuned LightGBM model using the best Optuna parameters.

# Start with Optuna best parameters.
best_lgbm_params = lgbm_study.best_params.copy()

# Add stable model settings.
best_lgbm_params.update({
    "n_estimators": 700 if FAST_MODE else 1500,
    "subsample": 0.85,
    "colsample_bytree": 0.85,
    "random_state": RANDOM_STATE,
    "n_jobs": -1,
    "verbose": -1
})

# Create the tuned LightGBM model.
best_lgbm_model = LGBMRegressor(**best_lgbm_params)

# Train the tuned model.
best_lgbm_model.fit(X_house_train_processed, y_house_train)

# Predict validation prices.
lgbm_tuned_pred = best_lgbm_model.predict(X_house_valid_processed)

# Calculate tuned LightGBM RMSE.
lgbm_tuned_rmse = calculate_rmse(y_house_valid, lgbm_tuned_pred)

print(f"Tuned LightGBM validation RMSE: ${lgbm_tuned_rmse:,.2f}")
```
